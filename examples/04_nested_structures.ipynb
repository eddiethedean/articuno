# 🗂️ Nested Structures

Handle complex nested dictionaries and lists.

```python
from articuno import infer_pydantic_model
import pandas as pd
```

## Nested Dictionaries

```python
data = [
    {
        "user_id": 1,
        "profile": {"name": "Alice", "age": 30},
        "tags": ["python", "ml"]
    }
]

Model = infer_pydantic_model(data, model_name="User")
instance = Model(**data[0])
print(f"User {instance.user_id}")
```

## Pandas Nested Columns

```python
df = pd.DataFrame({
    "id": [1, 2],
    "data": [{"x": 1}, {"x": 2}]
})

Model = infer_pydantic_model(df, model_name="Nested")
print("Fields:", list(Model.model_fields.keys()))
```
