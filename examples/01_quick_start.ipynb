# 🚀 Quick Start with Articuno

Learn the basics of articuno with simple examples.

```python
from articuno import df_to_pydantic, infer_pydantic_model
import polars as pl
import pandas as pd
```

## Polars DataFrame to Pydantic

```python
df = pl.DataFrame({
    "id": [1, 2, 3],
    "name": ["Alice", "Bob", "Charlie"],
    "score": [95.5, 88.0, 92.3],
    "active": [True, False, True]
})

instances = list(df_to_pydantic(df, model_name="User"))
print("First instance:", instances[0])
print(f"Created {len(instances)} total instances")
```

```python
# Or just get the model class
Model = infer_pydantic_model(df, model_name="UserModel")
print("\nModel name:", Model.__name__)
print("Fields:", list(Model.model_fields.keys()))
```

## Pandas DataFrame to Pydantic

```python
df_pd = pd.DataFrame({
    "user_id": [101, 102, 103],
    "username": ["alice", "bob", "charlie"],
    "points": [1500, 2300, 1800]
})

pandas_instances = list(df_to_pydantic(df_pd, model_name="PandasUser"))
print("Pandas instances:")
for inst in pandas_instances:
    print(f"  {inst}")
```

## Dict Iterables to Pydantic

```python
records = [
    {"id": 1, "name": "Alice", "email": "alice@example.com"},
    {"id": 2, "name": "Bob", "email": "bob@example.com"},
    {"id": 3, "name": "Charlie", "email": "charlie@example.com"}
]

instances = list(df_to_pydantic(records, model_name="Contact"))
print(f"Created {len(instances)} contact instances")
```

## Generator Behavior

`df_to_pydantic` returns a generator for memory efficiency.

```python
# Generator - memory efficient
instances_gen = df_to_pydantic(df, model_name="UserGen")
print("Type:", type(instances_gen))

# Collect all at once
instances_list = list(df_to_pydantic(df, model_name="UserList"))
print(f"Collected {len(instances_list)} instances as a list")
```
