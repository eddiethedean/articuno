# 🎯 Advanced Features

Explore advanced articuno capabilities.

```python
from articuno import infer_pydantic_model
```

## Generator Processing

```python
def generate_records():
    for i in range(100):
        yield {"id": i, "value": f"item_{i}"}

Model = infer_pydantic_model(generate_records(), model_name="Record", max_scan=10)
print("✓ Model from generator")
```

## Type Precedence

```python
data = [{"id": 1, "flag": True, "count": 10}]
Model = infer_pydantic_model(data, model_name="Typed")
instance = Model(**data[0])
print(f"flag is bool: {isinstance(instance.flag, bool)}")
```
