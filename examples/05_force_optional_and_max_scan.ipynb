# 🎛️ Force Optional & Max Scan

Control field optionality and scanning behavior.

```python
from articuno import infer_pydantic_model
import polars as pl
```

## Force Optional

```python
df = pl.DataFrame({"id": [1, 2], "name": ["A", "B"]})
Model = infer_pydantic_model(df, force_optional=True)
instance = Model(id=None, name=None)
print("✓ Force optional works:", instance)
```

## Max Scan

```python
records = [{"id": i, "value": f"item_{i}"} for i in range(1000)]
Model = infer_pydantic_model(records, model_name="Large", max_scan=50)
print("✓ Model created by scanning only 50 records")
```
