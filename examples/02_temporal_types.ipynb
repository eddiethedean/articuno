# 📅 Temporal Types Support

Articuno fully supports datetime, date, and timedelta types.

```python
from articuno import infer_pydantic_model, df_to_pydantic
from datetime import datetime, date, timedelta
import polars as pl
import pandas as pd
```

## Dict Iterables with Temporal Types

```python
events = [
    {
        "event_id": 1,
        "event_date": date(2024, 1, 15),
        "timestamp": datetime(2024, 1, 15, 10, 30),
        "duration": timedelta(hours=2, minutes=30)
    },
    {
        "event_id": 2,
        "event_date": date(2024, 1, 16),
        "timestamp": datetime(2024, 1, 16, 14, 0),
        "duration": timedelta(hours=1)
    }
]

EventModel = infer_pydantic_model(events, model_name="Event")
instances = list(df_to_pydantic(events, model=EventModel))
for event in instances:
    print(f"Event {event.event_id} on {event.event_date}")
```

## Polars Temporal

```python
df = pl.DataFrame({
    "id": [1, 2],
    "created": [datetime(2024, 1, 1), datetime(2024, 1, 2)],
    "date": [date(2024, 1, 1), date(2024, 1, 2)]
})

Model = infer_pydantic_model(df, model_name="PolarsTime")
print("✓ Polars temporal types work")
```

## Pandas Datetime

```python
df_pd = pd.DataFrame({
    "id": [1, 2],
    "timestamp": pd.to_datetime(["2024-01-15", "2024-01-16"])
})

Model = infer_pydantic_model(df_pd, model_name="PandasTime")
print("✓ Pandas datetime works")
```
