# 🌐 Processing API Responses

Convert API responses into typed Pydantic models.

```python
from articuno import infer_pydantic_model, df_to_pydantic
from datetime import datetime
```

## Simple API Response

```python
api_data = [
    {
        "status": "success",
        "data": {"user_id": 123, "username": "alice"},
        "timestamp": datetime(2024, 1, 15, 10, 30),
    }
]

APIModel = infer_pydantic_model(api_data, model_name="APIResponse")
instances = list(df_to_pydantic(api_data, model=APIModel))
print("Response:", instances[0].status)
```
