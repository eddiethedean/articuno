# 🗄️ Processing Database Results

Convert SQL query results into Pydantic models.

```python
from articuno import infer_pydantic_model, df_to_pydantic
from datetime import date
```

## SQL Query Results

```python
employees = [
    {
        "employee_id": 1,
        "name": "Alice",
        "department": "Engineering",
        "hire_date": date(2020, 1, 15),
    },
    {
        "employee_id": 2,
        "name": "Bob",
        "department": "Sales",
        "hire_date": date(2021, 3, 20),
    },
]

Employee = infer_pydantic_model(employees, model_name="Employee")
instances = list(df_to_pydantic(employees, model=Employee))
for emp in instances:
    print(f"{emp.name}: {emp.department}")
```
